# Tokenization


Tokenization is the process of splitting text into smaller units called tokens (such as words, sentences, or subwords) for further processing.

```python
import nltk
```

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
```

```python
a=" India is my country and all Indians are my brothers and sisters I love my country and I am proud of its rich and varied heritage.I shall always strive to be worthy of it."
```

```python
A=word_tokenize(a)
```

### stemming

Stemming is the process of reducing a word to its root or base form by chopping off prefixes/suffixes, often without ensuring the result is a real word (e.g., "running" → "run", "studies" → "studi").

```python
from nltk.stem import PorterStemmer
pst=PorterStemmer()
```

```python
pst.stem("teaching")
```

```python
pst.stem("growing")
```

```python
for i in A:
    print(pst.stem(i))
```

```python
pst.stem('universe')
```

```python
pst.stem('university')
```

```python
pst.stem("universal")
```

### Lemmatization

Lemmatization is the process of reducing a word to its dictionary base form (called a lemma) using vocabulary and grammar rules, ensuring the result is a real word (e.g., "running" → "run", "studies" → "study").

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
pst.stem('trouble')
```

```python
for i in A:
    print(lemmatizer.lemmatize(i))
```

### pos_tag

POS tagging (Part-of-Speech tagging) is the process of assigning a grammatical category (like noun, verb, adjective) to each word in a text based on its meaning and context.

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
for i in A:
    print(nltk.pos_tag([i]))
```

### Stopword

### Vectorizer

Vectorizer is a tool that converts text into numerical vectors (numbers) so that machine learning algorithms can process and analyze it (e.g., CountVectorizer, TfidfVectorizer).

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
a1="india is my country"
a2="All india is my brother"
a3="i love my country"
```

```python
vectorizer=CountVectorizer (stop_words='english')
vectors=vectorizer.fit_transform([a1,a2,a3])

feature_names=vectorizer.get_feature_names_out()
dense=vectors.todense()
result=pd.DataFrame(dense,columns=feature_names)
print(result)
```

### TF/IDF 

TF-IDF (Term Frequency-Inverse Document Frequency) is a statistical measure that evaluates how important a word is to a document within a collection of documents, giving higher weight to words that are frequent in a document but rare across all documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
 a1="india is my country"
a2="All india is my brother"
a3="i love my country"
```

```python
vectorizer=TfidfVectorizer(stop_words='english')
vectors=vectorizer.fit_transform([a1,a2,a3])
feature_name=vectorizer.get_feature_names_out()
dense=vectors.todense()
result=pd.DataFrame(dense,columns=feature_name)
print(result)
```
